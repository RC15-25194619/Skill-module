```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.collections import LineCollection
from matplotlib.colors import LinearSegmentedColormap

def save_bpm_sharp_wide_glow(
    csv_file_path,
    out_png_path="bpm_dashboard_sharp_wide_glow.png",
    width_in=16,
    height_in=4.5,
    dpi=120
):
    # --- 1. Read data ---
    try:
        df = pd.read_csv(csv_file_path)
    except UnicodeDecodeError:
        df = pd.read_csv(csv_file_path, encoding="gbk")

    # --- 2. Find BPM column ---
    cols = {col.lower(): col for col in df.columns}
    bpm_col = cols.get("bpm")

    if not bpm_col:
        bpm_col = next((col for col in df.columns if "heart" in col.lower()), None)

    if not bpm_col:
        raise ValueError("No 'bpm' or heart-related column found in the CSV file.")

    # --- 3. Clean data ---
    df[bpm_col] = pd.to_numeric(df[bpm_col], errors="coerce")
    df = df.dropna(subset=[bpm_col])

    if "time" in df.columns:
        df["time_obj"] = pd.to_datetime(df["time"], format="%H:%M:%S", errors="coerce")
        df = df.dropna(subset=["time_obj"])
        time_labels = df["time"].astype(str).values
    else:
        time_labels = np.array([str(i) for i in range(len(df))])

    if len(df) < 2:
        raise ValueError("Not enough valid data points to draw the BPM line.")

    x_raw = np.arange(len(df))
    y_data = df[bpm_col].values

    # --- 4. Create transparent canvas ---
    fig, ax = plt.subplots(figsize=(width_in, height_in), dpi=dpi)
    fig.patch.set_alpha(0)
    ax.set_facecolor("none")

    # --- 5. BPM colormap: match the image (deep cool purple -> violet -> magenta -> light lavender) ---
    bpm_colors = [
        "#3f3949",  # dark muted purple-gray
        "#5b5470",  # deep gray-violet
        "#7872a8",  # cool violet
        "#9d6cff",  # bright violet
        "#c13cff",  # magenta-violet
        "#f0d8ff"   # very light lavender
    ]
    cmap = LinearSegmentedColormap.from_list("bpm_line", bpm_colors, N=512)

    # --- 6. Interpolate for smoother line ---
    x_new = np.linspace(x_raw.min(), x_raw.max(), num=len(x_raw) * 10)
    y_new = np.interp(x_new, x_raw, y_data)

    # --- 7. Build gradient line segments ---
    points = np.array([x_new, y_new]).T.reshape(-1, 1, 2)
    segments = np.concatenate([points[:-1], points[1:]], axis=1)

    y_min = float(np.min(y_new))
    y_max = float(np.max(y_new))
    if y_min == y_max:
        y_max = y_min + 1.0

    norm = plt.Normalize(y_min, y_max)

    # --- 8. Draw wide glow ---
    glow_color = cmap(0.72)
    y_range = y_max - y_min
    if y_range == 0:
        y_range = 1.0

    base_spread = y_range * 0.12
    layers = 20
    max_spread_factor = 6.0

    for i in range(layers):
        factor = max_spread_factor * (1 - i / layers)
        current_spread = base_spread * factor
        alpha = 0.01 + (0.03 * (i / layers))

        ax.fill_between(
            x_new,
            y_new - current_spread,
            y_new + current_spread,
            color=glow_color,
            alpha=alpha,
            edgecolor="none",
            linewidth=0
        )

    # --- 9. Draw main gradient line ---
    lc_main = LineCollection(
        segments,
        cmap=cmap,
        norm=norm,
        linewidths=0.8,
        alpha=1.0
    )
    lc_main.set_array(y_new)
    ax.add_collection(lc_main)

    # --- 10. Axis limits ---
    ax.set_xlim(x_raw.min(), x_raw.max())
    margin = y_range * 0.6
    ax.set_ylim(y_min - margin, y_max + margin)

    # --- 11. Grid and axes styling: black ---
    ax.minorticks_on()
    ax.grid(True, which="major", color="black", linestyle="--", linewidth=0.5, alpha=0.15)
    ax.grid(True, which="minor", color="black", linestyle=":", linewidth=0.3, alpha=0.05)

    for spine in ax.spines.values():
        spine.set_color("black")
        spine.set_alpha(0.4)
        spine.set_linestyle("--")
        spine.set_linewidth(0.5)

    ax.set_ylabel("BPM", color="black", fontsize=10, fontname="monospace")
    ax.tick_params(axis="y", colors="black", labelsize=8)

    # --- 12. X-axis labels ---
    num_grids = 12
    tick_indices = np.linspace(0, len(df) - 1, num_grids, dtype=int)
    ax.set_xticks(tick_indices)

    target_labels = 6
    show_every = max(num_grids // target_labels, 1)

    labels = []
    for i, idx in enumerate(tick_indices):
        labels.append(time_labels[idx] if (i % show_every == 0) else "")

    ax.set_xticklabels(labels, rotation=0, ha="center", fontsize=9, color="black")
    ax.set_xlabel("Time", color="black", fontsize=12, fontname="monospace")
    ax.tick_params(axis="x", colors="black")

    # --- 13. Title ---
    plt.suptitle(
        "BPM Sharp Wide Glow",
        color="black",
        fontsize=16,
        fontname="monospace",
        y=0.97
    )

    # --- 14. Save transparent PNG ---
    plt.subplots_adjust(left=0.06, right=0.98, top=0.88, bottom=0.18)
    plt.savefig(
        out_png_path,
        dpi=dpi,
        transparent=True,
        bbox_inches="tight",
        pad_inches=0.05
    )
    plt.show()

# --- Run ---
file_path = "data6_final.csv"
save_bpm_sharp_wide_glow(file_path)
```
